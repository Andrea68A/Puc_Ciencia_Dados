##Criar os diretórios (DBFS) e limpar as bases de dados (DATABASE)

###Estrutura de pastas

dbfs:/mnt/netflix/

└──raw/  # Dados brutos

└──docs/  # Documentação

└──bronze/ # Camada bronze

└──silver/ # Camada silver

└──gold/ 	# Camada gold



###Imports necessários

```
import gc
```

###Limpando as bases de dados

```
%sql
-- Drop the database using IF EXISTS
-- DROP DATABASE <nome_database> CASCADE - deleta um database e suas tabelas 
DROP DATABASE IF EXISTS bronze CASCADE;
DROP DATABASE IF EXISTS silver CASCADE;
DROP DATABASE IF EXISTS gold CASCADE;
```

###Removendo as pastas

A opção recurse=True remove as subpastas (remoção recursiva)

```

dbutils.fs.rm("dbfs:/mnt/netflix", recurse=True)
dbutils.fs.rm("dbfs:/mnt/netflix/raw", recurse=True)
dbutils.fs.rm("dbfs:/mnt/netflix/docs",recurse=True)
dbutils.fs.rm("dbfs:/mnt/netflix/bronze", recurse=True)
dbutils.fs.rm("dbfs:/mnt/netflix/silver", recurse=True)
dbutils.fs.rm("dbfs:/mnt/netflix/gold", recurse=True)


```

```
%fs ls /mnt
```

###Criação de pastas no DBFS

Criar pastas no DBFS (Databricks File System), de acordo com as camadas raw, bronze, silver.

```
# Criar pastas no DBFS (Databricks File System)

netflix_dir = "dbfs:/mnt/netflix"
netflix_docs_dir = "dbfs:/mnt/netflix/docs"
netflix_raw_dir = "dbfs:/mnt/netflix/raw"
netflix_silver_dir = "dbfs:/mnt/netflix/silver/"
netflix_gold_dir = "dbfs:/mnt/netflix/gold"

dbutils.fs.mkdirs(netflix_dir)
dbutils.fs.mkdirs(netflix_docs_dir)
dbutils.fs.mkdirs(netflix_raw_dir)
dbutils.fs.mkdirs(netflix_silver_dir)
dbutils.fs.mkdirs(netflix_gold_dir)

```

###Limpeza dos dados em cache e coleta de lixo

```
# Limpar todos os dados em cache

spark.catalog.clearCache()

# clearCache() limpa o cache de todos os objetos em cache no SparkSession atual, liberando uma quantidade significativa de memória quando múltiplos DataFrames estão sendo reutilizados.


gc.collect()

#Comentário: Esse comando força o coletor de lixo a executar imediatamente, liberando a memória de objetos Python que não estão mais em uso.
```
